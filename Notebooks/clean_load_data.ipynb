## `load_cycle_data_10_cells` — column dictionary

Cleaned from `load_cycle_raw` (RealHyFC ~2300 h durability log; 10-cell, 280 cm² stack; ~1 Hz sampling).

**Abbreviations:** S = stack · Si = stack inlet · So = stack outlet · A = anode · C = cathode · CL = coolant loop · DPT = dew-point temperature · RH = relative humidity · `.set` = setpoint · j = current density. Gauge pressures are in kPag.

### Time
| Column | Meaning |
|---|---|
| timestamp | Wall-clock time of the sample (ISO text, ~1 Hz), kept from `DateTime`. |
| elapsed_h | Wall-clock hours since the first sample (2025-01-27 12:09:18). **Includes rig downtime**, so it is calendar time, not run time. |
| op_hours | Operating-hours counter (from `OpHrs`). Includes break-in + characterization and **matches the `OpHrs` in the polarization/EIS files** — use it to align load-cycle age with the characterization data. |
| op_hours2 | Secondary operating-hours counter (from `OpHrs2`); diverges from `op_hours` over the test (exact definition not documented by the dataset). |

### Current
| Column | Meaning |
|---|---|
| I.S.set [A] / I.S.act [A] | Stack current — setpoint / actual. |
| j.S.act [A/cm2] | Stack current density (I.S.act / 280 cm²). |
| I.S.set.internal [A] | Internal controller current setpoint. |
| I.S.MinGasFlow [A] | Current equivalent of the minimum gas flow (floor below which stoichiometry cannot be held). |

### Reactant supply — stoichiometry & humidity
| Column | Meaning |
|---|---|
| Stoic.S.A [-] / .set | Anode stoichiometry — actual / setpoint. |
| Stoic.S.C [-] / .set | Cathode stoichiometry — actual / setpoint. |
| RH.Si.A [%] / RH.Si.A.CV [%] | Anode-inlet relative humidity (and its control/variation channel). |
| RH.Si.C [%] / RH.Si.C.CV [%] | Cathode-inlet relative humidity (and its control/variation channel). |

### Voltages
| Column | Meaning |
|---|---|
| MAX(`U.S.MaxCell [V]`) | Highest single-cell voltage among the channels. |
| U.S.AveCell [V] | Average cell voltage. |
| MIN(`U.S.MinCell [V]`) | Lowest single-cell voltage. |
| U1…U11 [V] | Individual single-cell voltages (`U6` removed — dead channel, ~0 V). |
| MIN(`U.S [V]`) | Minimum **stack** voltage (all cells summed) — a different quantity from the min *cell* voltage above. NaN in logger versions that did not record it (e.g. `Loadcycle_51-70`). |

### Flows
| Column | Meaning |
|---|---|
| F.Si.H.A [nl/min] / .set | Anode-inlet H₂ flow. |
| F.Si.N.A [nl/min] / .set | Anode-inlet N₂ flow (fuel is 80 % H₂ / 20 % N₂). |
| F.Si.Air.C [nl/min] / .set | Cathode-inlet air flow. |
| F.Si.A.total [nl/min] / .set | Total anode-inlet flow. |
| F.Si.C.total [nl/min] / .set | Total cathode-inlet flow. |
| F.Si.CL [l/min] / .set | Coolant flow. |

### Pressures (kPag)
| Column | Meaning |
|---|---|
| p.S.A.set | Anode pressure setpoint. |
| p.Si.A / p.So.A | Anode inlet / outlet pressure. |
| p.S.C.set | Cathode pressure setpoint. |
| p.Si.C / p.So.C | Cathode inlet / outlet pressure. |
| p.Si.CL / p.So.CL | Coolant inlet / outlet pressure. |

### Temperatures & dew points (°C)
| Column | Meaning |
|---|---|
| T.Si.A [°C] / .set | Anode-inlet temperature. |
| T.So.A [°C] | Anode-outlet temperature. |
| T.Si.C [°C] / T.Si.C.set | Cathode-inlet temperature / setpoint. |
| T.So.C [°C] | Cathode-outlet temperature. |
| T.Si.CL [°C] / .set | Coolant-inlet temperature. |
| T.So.CL [°C] | Coolant-outlet temperature. |
| DPT.Si.A [°C] / .set | Anode-inlet dew point (sets anode humidity). |
| DPT.Si.C [°C] / .set | Cathode-inlet dew point (sets cathode humidity). |

### Operating point
| Column | Meaning |
|---|---|
| operation_point | Index of the current step within the RealHyFC load cycle. |

**Dropped in cleaning:** `CalHrs [h]` (sparse — only 2 files, a per-segment counter, redundant with `timestamp`), `Number of cells` (constant = 10, encoded in the table name), `U6 [V]` (dead channel), `source_file` (provenance). `DateTime` → `timestamp`; `OpHrs` → `op_hours`; `OpHrs2` → `op_hours2`.


```python
# --- Clean load_cycle_raw -> load_cycle_data_10_cells (chunked) ---
# Reads the 8.6 M-row raw table in chunks so it never has to fit in memory at once.
import sqlite3
import pandas as pd
from contextlib import closing

NEW_TABLE = "load_cycle_data_10_cells"
CHUNK     = 500_000
START     = pd.Timestamp("2025-01-27 12:09:18.005")   # first sample -> t0 for elapsed_h

# removed from the clean table (see the markdown dictionary above):
DROP = ["DateTime", "CalHrs [h]", "OpHrs [h]", "OpHrs2 [h]",
        "Number of cells", "U6 [V]", "source_file"]

def clean_chunk(df):
    ts = pd.to_datetime(df["DateTime"], format="%Y-%m-%d %H:%M:%S.%f")
    out = df.drop(columns=DROP)
    out.insert(0, "timestamp", df["DateTime"])                        # wall-clock (ISO text)
    out.insert(1, "elapsed_h", (ts - START).dt.total_seconds() / 3600)  # wall-clock hours since t0
    out.insert(2, "op_hours",  df["OpHrs [h]"])                       # operating hours (aligns w/ characterization)
    out.insert(3, "op_hours2", df["OpHrs2 [h]"])                      # secondary operating counter
    return out

# One connection, keyset pagination on rowid: each chunk is a complete, finished SELECT,
# so no read cursor stays open (holding a SHARED lock) while to_sql commits its inserts.
QUERY = "SELECT rowid AS _rid, * FROM load_cycle_raw WHERE rowid > ? ORDER BY rowid LIMIT ?"

with closing(sqlite3.connect(DB_path, timeout=30)) as conn:
    conn.execute(f'DROP TABLE IF EXISTS "{NEW_TABLE}"')
    conn.commit()
    total, last_rid, i = 0, 0, 0
    while True:
        chunk = pd.read_sql_query(QUERY, conn, params=(last_rid, CHUNK))
        if chunk.empty:
            break
        last_rid = int(chunk["_rid"].iloc[-1])
        clean_chunk(chunk.drop(columns="_rid")).to_sql(NEW_TABLE, conn, if_exists="append", index=False)
        conn.commit()
        total += len(chunk)
        print(f"  chunk {i:>2}: {total:>10,} rows written")
        i += 1
print(f"Done: {total:,} rows -> {NEW_TABLE}")
```
